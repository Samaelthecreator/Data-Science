# Ingeniera de datos

Para trabajar en modelos predictivos, necesitamos de primero hacer la extracción y curado de la información, para esto será importante determinar el camino de este trabajo, la metodologia a seguir será:

- Identificación de fuentes de la información: determinamos de donde vendrá la información y si los proveedores de la información tienen autoridad para ofertarla.
- Extracción y recolección: determinar los métodos de extracción (web scrapping, api, manual, etc) formato de datos (hojas de excel, archivos de texto, jsons, data lakes, etc) y  el almacenamiento (plataformas en la nube o en el dispositivo local).
- Analisis y esquematización: Analizar los datos y su contenido (información) para determinar la mejor estructura para darle contexto a la información (relacional, documentos, diccionarios, grafos, etc).

- Ordenamiento y limpieza: a partir de identificar la forma mas idónea de trabajar con los datos, comenzamos con el ordenamiento de la información y la limpieza (datos perdidos, formatos incorrectos, ingeniera de caracteristicas, etc).

- Validación y almacenaje: posteriormente a la limpieza y estructuración, validar los datos a través de indicadores de calidad y creación de metadata son importantes para tener la información lista para el proceso de modelaje (Ciencia de datos).

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import time
import random

class Inmuebles24Scraper:
    def __init__(self):
        self.base_url = "https://www.inmuebles24.com/casas-en-venta-en-{}-pagina-{}.html"
        self.headers = {
            "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/119.0.0.0 Safari/537.36",
            "Accept-Language": "es-MX,es;q=0.9",
            "Referer": "https://www.google.com/"
        }
        self.data = []

    def fetch_page(self, ciudad, pagina):
        url = self.base_url.format(ciudad, pagina)
        print(f"Extrayendo: {url}")
        
        try:
            # Usamos un timeout para no quedar colgados
            response = requests.get(url, headers=self.headers, timeout=15)
            
            if response.status_code == 200:
                return BeautifulSoup(response.text, 'html.parser')

            elif response.status_code == 403:
                print(f"Error 403 en: {url}")
                print("--- Detalles del Bloqueo ---")
                print(f"Server: {response.headers.get('Server')}") # Probablemente diga 'cloudflare'
                print(f"CF-Chl-Bypass: {response.headers.get('cf-chl-bypass')}") # Firma de desafío
                # Si ves 'cf-ray', es confirmación total de Cloudflare
                return None

        except Exception as e:
            print(f"Error en la petición: {e}")
            return None

    def parse_data(self, soup, ciudad):
        # Inmuebles24 suele envolver los datos en artículos o divs con clases específicas
        # Nota: Estas clases cambian frecuentemente, es un punto de enseñanza para el curso
        postings = soup.find_all('div', {'data-qa': 'posting-container'})
        
        for post in postings:
            try:
                item = {
                    'titulo': post.find('div', {'class': 'sc-1v1y961-1'}).text.strip() if post.find('div', {'class': 'sc-1v1y961-1'}) else "N/A",
                    'precio': post.find('div', {'data-qa': 'POSTING_CARD_PRICE'}).text.strip() if post.find('div', {'data-qa': 'POSTING_CARD_PRICE'}) else "Consultar",
                    'ubicacion': post.find('div', {'data-qa': 'POSTING_CARD_LOCATION'}).text.strip() if post.find('div', {'data-qa': 'POSTING_CARD_LOCATION'}) else "N/A",
                    'descripcion': post.find('div', {'data-qa': 'POSTING_CARD_DESCRIPTION'}).text.strip() if post.find('div', {'data-qa': 'POSTING_CARD_DESCRIPTION'}) else "N/A",
                    'ciudad_busqueda': ciudad
                }
                self.data.append(item)
            except Exception as e:
                continue

    def run(self, ciudades, max_paginas=3):
        for ciudad in ciudades:
            for i in range(1, max_paginas + 1):
                soup = self.fetch_page(ciudad, i)
                if soup:
                    self.parse_data(soup, ciudad)
                
                # Timer aleatorio para evitar detección (3 a 7 segundos)
                wait_time = random.uniform(3, 7)
                print(f"Esperando {wait_time:.2f} segundos...")
                time.sleep(wait_time)
        
        return pd.DataFrame(self.data)

# --- USO DEL SCRIPT ---
# Lista de ciudades (asegurar que el formato coincida con el slug de la URL)
lista_ciudades = ["queretaro", "guadalajara", "monterrey"]

scraper = Inmuebles24Scraper()
df_inmuebles = scraper.run(lista_ciudades, max_paginas=2)

# Presentación de la información
print(df_inmuebles)

# Guardar resultado
df_inmuebles.to_csv("datos_inmuebles_crudos.csv", index=False)
print("Extracción finalizada.")
```

```python
import cloudscraper
from bs4 import BeautifulSoup
import json
import re

def extraer_datos_robusto(ciudad):
    # Creamos un scraper que imita un navegador real (Chrome en Windows)
    scraper = cloudscraper.create_scraper(
        browser={
            'browser': 'chrome',
            'platform': 'windows',
            'desktop': True
        }
    )
    
    url = f"https://www.inmuebles24.com/casas-en-venta-en-{ciudad}.html"
    response = scraper.get(url)
    
    if response.status_code == 200:
        # TRUCO MAESTRO: Buscar el JSON oculto en el HTML
        # Inmuebles24 suele guardar todo en window.__PRELOADED_STATE__
        pattern = re.compile(r'window\.__PRELOADED_STATE__\s*=\s*(\{.*?\});', re.DOTALL)
        match = pattern.search(response.text)
        
        if match:
            json_data = json.loads(match.group(1))
            # Aquí tienes TODO el dataset sin haber scrapeado un solo DIV
            # Es una mina de oro de información estructurada
            return json_data
    else:
        print(f"Sigo bloqueado con error {response.status_code}")
        return None

# Ejemplo de uso
datos = extraer_datos_robusto("queretaro-qro")
```
